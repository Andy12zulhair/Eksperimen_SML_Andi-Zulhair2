```python
import os
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

print('Libraries loaded successfully!')
```

## 1. Memuat Dataset Mentah (Raw Dataset)

```python
raw_path = os.path.join('..', 'namadataset_raw', 'diabetes_risk_raw.csv')
df = pd.read_csv(raw_path)

print('Dimensi Dataset Mentah:', df.shape)
df.head()
```

## 2. Eksplorasi Data (EDA) & Pengecekan Missing Value

```python
# Cek tipe data dan non-null count
df.info()
```

```python
# Pengecekan missing values secara eksplisit
print('Jumlah Missing Value per Kolom:')
print(df.isna().sum())
```

```python
# Pengecekan duplikasi baris data
print('Jumlah Duplikat:', df.duplicated().sum())
```

```python
# Distribusi label target diabetes_risk
print('Distribusi Target (diabetes_risk):')
print(df['diabetes_risk'].value_counts())
```

## 3. Data Cleaning (Imputasi Missing Values)
Pada fitur alcohol_consumption, data yang tidak terisi (missing) diisi dengan kategori 'None'.

```python
df['alcohol_consumption'] = df['alcohol_consumption'].fillna('None')
print('Missing value setelah imputasi:', df.isna().sum().sum())
```

## 4. Feature Encoding (Kategorikal ke Numerik)

```python
# Gender encoding
gender_map = {'Female': 0, 'Male': 1, 'Other': 2}
df['gender'] = df['gender'].map(gender_map).fillna(1)

# City encoding
cities = [
    'Mumbai', 'Delhi', 'Bangalore', 'Hyderabad', 'Ahmedabad', 'Chennai',
    'Kolkata', 'Surat', 'Pune', 'Jaipur', 'Lucknow', 'Kanpur',
    'Nagpur', 'Indore', 'Thane', 'Bhopal', 'Visakhapatnam', 'Patna'
]
city_map = {city: i for i, city in enumerate(cities)}
df['city'] = df['city'].map(city_map).fillna(0)

# Physical activity level
act_map = {'Sedentary': 0, 'Moderate': 1, 'Active': 2}
df['physical_activity_level'] = df['physical_activity_level'].map(act_map).fillna(1)

# Diet type
diet_map = {'Balanced': 0, 'Vegan': 1, 'Vegetarian': 2, 'Non-Vegetarian': 3}
df['diet_type'] = df['diet_type'].map(diet_map).fillna(0)

# Smoking status
smoke_map = {'Current': 0, 'Former': 1, 'Never': 2}
df['smoking_status'] = df['smoking_status'].map(smoke_map).fillna(2)

# Alcohol consumption
alc_map = {'None': 0, 'Moderate': 1, 'High': 2}
df['alcohol_consumption'] = df['alcohol_consumption'].map(alc_map).fillna(0)

# Income bracket
inc_map = {'Low': 0, 'Medium': 1, 'High': 2}
df['income_bracket'] = df['income_bracket'].map(inc_map).fillna(1)

print('Encoding selesai. Tipe data kolom:')
print(df.dtypes)
```

## 5. Feature Scaling (Standardization)
Melakukan standardisasi (z-score scaling) pada seluruh fitur prediktor agar performa model optimal.

```python
target_col = 'diabetes_risk'
feature_cols = [c for c in df.columns if c != target_col]

scaler = StandardScaler()
df[feature_cols] = scaler.fit_transform(df[feature_cols])

print('Hasil Statistik Deskriptif Pasca-Standarisasi:')
df.describe()
```

## 6. Menyimpan Dataset Hasil Preprocessing

```python
clean_output_dir = 'namadataset_preprocessing'
os.makedirs(clean_output_dir, exist_ok=True)
clean_output_path = os.path.join(clean_output_dir, 'diabetes_risk_clean.csv')

df.to_csv(clean_output_path, index=False)
print('Dataset preprocessing berhasil disimpan ke:', clean_output_path)
print('Ukuran akhir dataset:', df.shape)
```
